# **Data Cleaning and Preprocessing**
# Dataset: Churn Dataset (BigML)

  This dataset contains telecom customer data with 2666 rows and 20 columns, including customer usage patterns, account information and whether the customer churned or not.

**Objective:**
1) Load the dataset using pandas
2) Identify and handle missing values
3) Remove duplicate rows and standardize inconsistent data formats

**Problems**
1) Dataset didn't contain missing values,duplicates and inconsistent data formats. So inconsistencies were artificially introduced to simulate real-world dirty data, including missing values, duplicate rows, and inconsistent formatting.

**Findings**


*   The dataset contained 29 missing values. Seven in Voice mail plan, eleven in Total day minutes column, and eleven in Total night minutes column.

*   A total of 50 duplicates was removed.

*   The Column - International Plan had inconsistent text formatting, where some values were lowercase while others were uppercase. These were all standardized to proper case..


**Tools Used**
- Python 3
- Pandas — for data loading, cleaning and manipulation
- NumPy — for introducing missing values (np.nan).


**Findings**

  After cleaning, the dataset was free of missing values, duplicates and inconsistent formats, making it ready for exploratory data analysis.

I began by importing the pandas library, which is the primary tool used for data manipulation and analysis in Python.

```python
# Importing pandas library for data manipulation.
import pandas as pd
```

Next I loaded the dataset into a pandas DataFrame. The dataset is stored in Google Drive and is read using the read_csv() function.

```python
# Load the dataset from Google Drive into a pandas DataFrame
df = pd.read_csv('/content/drive/MyDrive/Churn Prdiction Data/churn-bigml-80.csv')
```

Before cleaning the data, I first explored its basic structure. The info() function gives mean overview of the dataset including the number of rows, column names, data types and whether any null values exist.

```python
# Display basic information about the dataset
df.info()
```

I displayed the full dataset to get a visual overview of its contents, structure and values.


```python
# Display the full dataframe
df
```

I added NaN values to two numerical columns and one categorical column.

```python
# Import numpy to access NaN (missing value representation)
import numpy as np

# Introduce missing values in numerical columns
df.loc[50:60   , 'Total day minutes']   = np.nan

# Introduce missing values in another numerical column
df.loc[0:10    , 'Total night minutes']  = np.nan

# Introduce missing values in a categorical column
df.loc[34:40   , 'Voice mail plan']       = np.nan

```

I artificially introduced duplicate rows by selecting a range of existing rows and appending them back to the bottom of the dataframe

```python
# Select rows 1000 to 1050 to use as duplicates
duplicates = df.iloc[1000:1050]

#Append duplicate rows to the bottom of the dataframe
df = pd.concat([df, duplicates], ignore_index=True)
```

I introduced inconsistent formatting in the International plan column by using different cases and extra spaces to simulate real-world data entry errors.

```python
# Introduce lowercase inconsistency
df.loc[2600:2630, 'International plan'] = 'yes'

# Introduce Uppercase with extra space inconsistency
df.loc[350:370, 'International plan']   = ' YES '
```

I checked for missing values in each column to confirm they exist and identify which columns are affected

```python
# Count missing values in each column
df.isnull().sum()
```

I handled missing values differently depending on the column type. For numerical columns I filled with the mean since it represents the average value. For categorical columns I filled with the mode which is the most frequently occurring value

```python
# Fill missing categorical values with the most frequent value (mode)
df['Voice mail plan'] = df['Voice mail plan'].fillna(df['Voice mail plan'].mode()[0])

# Fill missing numerical values with the column mean
df['Total day minutes'] = df['Total day minutes'].fillna(df['Total day minutes'].mean())

# Fill missing numerical values with the column mean
df['Total night minutes'] = df['Total night minutes'].fillna(df['Total night minutes'].mean())
```

I ran the missing values check again to confirm that all NaN values have been successfully handled.

```python
# Verify missing values have been handled
df.isnull().sum()
```

I checked how many duplicate rows exist in the dataset before removing them.

```python
# Count total number of duplicate rows
print(df.duplicated().sum())
```

I displayed the actual duplicate rows so I can visually confirm which rows are repeated before removing them.

```python
# Display the duplicate rows
df[df.duplicated()]
```

I removed all duplicate rows from the dataset keeping only the first occurrence of each row

```python
# Remove all duplicate rows
df.drop_duplicates(inplace=True)
```

I verified that all duplicates have been successfully removed

```python
# Confirm duplicates have been removed
print(df.duplicated().sum())
```

I checked all categorical columns for inconsistent formatting by looking at their unique values and how many times each appears.

```python
# Check unique values in categorical columns
print(df['International plan'].value_counts())
print(df['Voice mail plan'].value_counts())
print(df['State'].value_counts())
print(df['Churn'].value_counts())
```

I standardized the International plan column by removing extra spaces with strip() and converting all values to title case with title().

```python
# Remove extra spaces and standardize casing in International plan column
df['International plan'] = df['International plan'].str.strip().str.title()
```

I verified that the International plan column now has consistent formatting by checking its unique values again

```python
# Confirm formatting is now consistent
print(df['International plan'].value_counts())
```
